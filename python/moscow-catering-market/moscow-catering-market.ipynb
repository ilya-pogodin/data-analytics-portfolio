## 2. Предобработка данных

Подготовьте данные к исследовательскому анализу:

- Изучите корректность типов данных и при необходимости проведите их преобразование.

```python
# исходные типы данных
mem_before = df.memory_usage(deep=True).sum() / 1024**2
display(df.dtypes.to_frame('тип'))

# все значения seats целые — вещественный тип возник только из-за пропусков
print('Все значения seats целые:', (df['seats'].dropna() % 1 == 0).all())
print('Диапазон rating:', df['rating'].min(), '—', df['rating'].max())
print('Уникальные значения chain:', sorted(df['chain'].unique()))
print('Объём памяти до преобразования, МБ:', round(mem_before, 2))
```

```python
# ценовая категория — порядковая переменная, задаём правильный порядок уровней
price_order = ['низкие', 'средние', 'выше среднего', 'высокие']
df['price'] = pd.Categorical(df['price'], categories=price_order, ordered=True)

# номинальные признаки переводим в тип category: меньше памяти и удобнее группировки
for col in ['category', 'district']:
    df[col] = df[col].astype('category')

# флаг сетевого заведения достаточно хранить в int8
df['chain'] = df['chain'].astype('int8')

mem_after = df.memory_usage(deep=True).sum() / 1024**2
display(df.dtypes.to_frame('тип'))
print('Объём памяти: было %.2f МБ, стало %.2f МБ' % (mem_before, mem_after))
```

```python
# общее распределение количества посадочных мест
seats = df['seats'].dropna()
print('Количество мест указано у %d заведений (%.1f%% от всех)' % (len(seats), len(seats) / len(df) * 100))
display(df['seats'].describe().round(1).to_frame('посадочные места'))

q1, q3 = seats.quantile([0.25, 0.75])
iqr_limit = q3 + 1.5 * (q3 - q1)
print('Верхняя граница «усов» по правилу 1.5 IQR: %.0f мест' % iqr_limit)
print('Заведений выше этой границы: %d (%.1f%% от заполненных)' %
      ((seats > iqr_limit).sum(), (seats > iqr_limit).mean() * 100))

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))

sns.histplot(seats, bins=60, color='#2a7f83', ax=axes[0])
axes[0].axvline(seats.median(), color='crimson', linestyle='--',
                label='медиана — %.0f мест' % seats.median())
axes[0].set_title('Распределение количества посадочных мест')
axes[0].set_xlabel('Количество посадочных мест, шт.')
axes[0].set_ylabel('Количество заведений, шт.')
axes[0].legend()

sns.boxplot(x=seats, color='#2a7f83', ax=axes[1])
axes[1].set_title('Посадочные места: выбросы видны справа')
axes[1].set_xlabel('Количество посадочных мест, шт.')

plt.tight_layout()
plt.show()
```

```python
# phi_k подходит для смешанных данных (числовые + категориальные признаки),
# в окружении тренажёра библиотеки нет — устанавливаем
!pip install -q phik

import phik
print('Версия phik:', phik.__version__)
```

```python
# признаки из задания берём в исходном виде: phi_k не требует кодирования категорий,
# поэтому категория заведения и округ остаются одним столбцом, а не десятком дамми
features = pd.DataFrame({
    'рейтинг': df['rating'],
    'категория': df['category'].astype(str),
    'округ': df['district'].astype(str).str.replace(' административный округ', '', regex=False),
    'сетевое заведение': df['chain'].map({0: 'нет', 1: 'да'}),
    'посадочные места': df['seats'],
    'ценовая категория': df['price'].astype(str).replace('nan', np.nan),
    'круглосуточно': df['is_24_7'].map({False: 'нет', True: 'да'})
})

# interval_cols — непрерывные признаки, phi_k разбивает их на интервалы автоматически
phik_corr = features.phik_matrix(interval_cols=['рейтинг', 'посадочные места'])
display(phik_corr.round(2))

fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(phik_corr, annot=True, fmt='.2f', cmap='YlGnBu', vmin=0, vmax=1,
            square=True, linewidths=0.5,
            cbar_kws={'label': 'коэффициент phi_k (от 0 до 1)'}, ax=ax)
ax.set_title('Матрица корреляции phi_k: рейтинг и характеристики заведения')
ax.set_xlabel('')
ax.set_ylabel('')
plt.xticks(rotation=30, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# ранжируем связи рейтинга со всеми признаками
rating_phik = phik_corr['рейтинг'].drop('рейтинг').sort_values(ascending=False)
display(rating_phik.round(3).to_frame('phi_k с рейтингом'))

fig, ax = plt.subplots(figsize=(9, 4.5))
sns.barplot(x=rating_phik.values, y=list(rating_phik.index), palette='crest_r', ax=ax)
for i, v in enumerate(rating_phik):
    ax.text(v + 0.005, i, '%.3f' % v, va='center', fontsize=9)
ax.set_title('Сила связи рейтинга с характеристиками заведения (phi_k)')
ax.set_xlabel('Коэффициент phi_k: 0 — связи нет, 1 — функциональная зависимость')
ax.set_ylabel('Признак')
ax.set_xlim(0, 0.3)
plt.show()

print('Самая сильная связь рейтинга: %s (phi_k = %.3f)' % (rating_phik.index[0], rating_phik.iloc[0]))
```

```python
# проверяем самую сильную связь: рейтинг и ценовая категория
price_known = df.dropna(subset=['price']).copy()
price_known['price_rank'] = price_known['price'].cat.codes

rho, p_spearman = stats.spearmanr(price_known['price_rank'], price_known['rating'])
print('Ранговая корреляция Спирмена: rho = %.3f, p-value = %.2g (n = %d)' % (rho, p_spearman, len(price_known)))

groups = [g['rating'].values for _, g in price_known.groupby('price') if len(g) > 0]
h_stat, p_kruskal = stats.kruskal(*groups)
print('Критерий Краскела — Уоллиса: H = %.1f, p-value = %.2g' % (h_stat, p_kruskal))

display(price_known.groupby('price')['rating'].agg(['count', 'mean', 'median']).round(2)
        .rename(columns={'count': 'заведений', 'mean': 'средний рейтинг', 'median': 'медиана'}))

# вторая по силе связь: расположение в ЦАО
cao_rating = df.loc[df['district'] == CAO, 'rating']
other_rating = df.loc[df['district'] != CAO, 'rating']
u_stat, p_mw = stats.mannwhitneyu(cao_rating, other_rating)
print()
print('Средний рейтинг в ЦАО: %.2f (n = %d)' % (cao_rating.mean(), len(cao_rating)))
print('Средний рейтинг в остальных округах: %.2f (n = %d)' % (other_rating.mean(), len(other_rating)))
print('Критерий Манна — Уитни: p-value = %.2g' % p_mw)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))

sns.boxplot(data=price_known, x='price', y='rating', palette='crest', ax=axes[0])
axes[0].set_title('Рейтинг заведений по ценовым категориям')
axes[0].set_xlabel('Ценовая категория')
axes[0].set_ylabel('Рейтинг')

zones = df.assign(zone=np.where(df['district'] == CAO, 'ЦАО', 'остальные округа'))
sns.boxplot(data=zones, x='zone', y='rating', palette='crest', ax=axes[1])
axes[1].set_title('Рейтинг: ЦАО и остальные округа')
axes[1].set_xlabel('')
axes[1].set_ylabel('Рейтинг')

plt.tight_layout()
plt.show()
```

### Расчёты для итоговых рекомендаций

Числа, на которые опираются рекомендации шага 4: насыщенность округов дорогими заведениями и профиль сетевых кофеен эконом-сегмента — прямых конкурентов при выборе недорогой ниши.

```python
# 1) доля заведений с ценами «выше среднего» и «высокими» по округам
expensive = df.dropna(subset=['price']).copy()
expensive['district_short'] = expensive['district'].astype(str).str.replace(' административный округ', '', regex=False)
expensive['is_expensive'] = expensive['price'].isin(['выше среднего', 'высокие'])

share_expensive = (expensive.groupby('district_short')
                   .agg(с_известной_ценой=('is_expensive', 'size'), дорогих=('is_expensive', 'sum')))
share_expensive['доля дорогих, %'] = (share_expensive['дорогих'] / share_expensive['с_известной_ценой'] * 100).round(1)
share_expensive = share_expensive.sort_values('доля дорогих, %', ascending=False)

print('Доля заведений с ценами «выше среднего» и «высокими» по округам:')
display(share_expensive)

# 2) сетевые кофейни эконом-сегмента — прямые конкуренты при выборе недорогой ниши
competitors = ['cofix', 'one price coffee', 'кофепорт', 'шоколадница']
comp = (df[(df['chain'] == 1) & (df['name_norm'].isin(competitors))]
        .groupby('name_norm')
        .agg(точек=('id', 'count'),
             средний_рейтинг=('rating', 'mean'),
             медианная_цена_капучино=('middle_coffee_cup', 'median'))
        .sort_values('точек', ascending=False)
        .round(2))

print('Сетевые кофейни, с которыми придётся конкурировать в недорогом сегменте:')
display(comp)
```

**Базовый сценарий: кофейня-пекарня на 50–80 посадочных мест в ценовом сегменте «выше среднего».**
